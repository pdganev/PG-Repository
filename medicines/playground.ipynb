```python
df_2 = pd.read_csv('A_Z_medicines_dataset_of_India.csv')
df_2.shape
```

```python
#generic preprocessing
def process_text(text):
    # Handle NaN/float values by converting to string
    if pd.isna(text):
        return ''
    # Convert to string and process
    text = str(text)
    text = text.lower()
    text = text.replace(' ', '_')
    text = text.replace('-', '_')
    text = text.replace('/', '_')
    text = text.replace('(', '')
    text = text.replace(')', '')
    return text

columns_to_process = ['name', 'substitute0', 'substitute1', 'substitute2',
       'substitute3', 'substitute4', 'sideEffect0', 'sideEffect1',
       'sideEffect2', 'sideEffect3', 'sideEffect4', 'sideEffect5',
       'sideEffect6', 'sideEffect7', 'sideEffect8', 'sideEffect9',
       'sideEffect10', 'sideEffect11', 'sideEffect12', 'sideEffect13',
       'sideEffect14', 'sideEffect15', 'sideEffect16', 'sideEffect17',
       'sideEffect18', 'sideEffect19', 'sideEffect20', 'sideEffect21',
       'sideEffect22', 'sideEffect23', 'sideEffect24', 'sideEffect25',
       'sideEffect26', 'sideEffect27', 'sideEffect28', 'sideEffect29',
       'sideEffect30', 'sideEffect31', 'sideEffect32', 'sideEffect33',
       'sideEffect34', 'sideEffect35', 'sideEffect36', 'sideEffect37',
       'sideEffect38', 'sideEffect39', 'sideEffect40', 'sideEffect41', 'use0',
       'use1', 'use2', 'use3', 'use4', 'Chemical Class', 'Habit Forming',
       'Therapeutic Class', 'Action Class']

df = df.fillna('missing')
for column in columns_to_process:
    df[column] = df[column].apply(process_text)
df = df.drop(columns=['id'])

df_2 = df_2.rename(columns={'price(₹)' : 'price'})
df_2 = df_2.drop(columns=['id'])
columns_to_process = ['name', 'price', 'Is_discontinued', 'manufacturer_name',
       'type', 'pack_size_label', 'short_composition1', 'short_composition2']
df_2 = df_2.fillna('missing')
for column in columns_to_process:
    df_2[column] = df_2[column].apply(process_text) 

merged=df.merge(df_2, on=['name'])
merged = merged.drop_duplicates(subset='name')
merged.shape
```
